```python
import duckdb
```

```python
# Connect to a new DuckDB file (creates weather.duckdb if it doesn't exist)
con = duckdb.connect("weather.duckdb")

# Create a simple table
con.execute("""
CREATE TABLE weather (
    id INTEGER,
    city VARCHAR,
    temperature FLOAT,
    condition VARCHAR
)
""")
```

```python
# Insert some rows
con.execute("INSERT INTO weather VALUES (1, 'Seattle', 55.2, 'Cloudy')")
con.execute("INSERT INTO weather VALUES (2, 'Phoenix', 88.9, 'Sunny')")

# Query it
df = con.execute("SELECT * FROM weather").df()
print(df)
```

```python
con.close()
```
